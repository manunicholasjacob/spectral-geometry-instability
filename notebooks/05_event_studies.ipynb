## Event Statistics

```python
print("Event Statistics for SGI:")
for event_name, series_stats in event_results['statistics'].items():
    if 'sgi' in series_stats:
        stats = series_stats['sgi']
        print(f"\n{event_name}:")
        print(f"  Event value: {stats['event_value']:.4f}")
        print(f"  Pre-event mean: {stats['pre_mean']:.4f}")
        print(f"  Post-event mean: {stats['post_mean']:.4f}")
        print(f"  Event z-score: {stats['event_zscore']:.2f}" if stats['event_zscore'] else "  Event z-score: N/A")
```
